### DataFrame

A DataFrame is a two-dimensional table in Pandas that stores data in rows and columns, similar to an Excel spreadsheet.

```python
import pandas as pd
df = pd.DataFrame([11,22,33])
print(df)
```

```python
import pandas as pd
df = pd.DataFrame([11,22,33],columns=['col_name'])
print(df)
```

```python
print(type(df))
```

```python
import pandas as pd

data = {
    'Name': ['Yangla', 'Sonam', 'Gyajola'],
    'Age': [24, 27, 22],
    'Salary': [20000, 30000, 40000]
}

df = pd.DataFrame(data)

print(df)
```

### Filter DataFrame

```python
import pandas as pd

data = {
    'Name': ['Yangla', 'Sonam', 'Gyajola'],
    'Age': [24, 27, 22],
    'Salary': [20000, 30000, 40000]
}

df = pd.DataFrame(data)
df
```

```python
df
```

```python
df['Age'] >= 23
```

```python
df_age_filter = df['Age'] >= 23
```

```python
df_age_filter
```

```python
df[(df['Age'] >= 23) & (df['Salary'] >=20000)]
```

```python
df.where(df['Age'] >= 23)
```

```python
df.where(df['Age'] >= 23,other = 'not eligible')
```

```python
import pandas as pd

data = {
    'Name': ['Yangla', 'Sonam', 'Gyajola'],
    'Age': [24, 27, 22],
    'Salary': [20000, 30000, 40000]
}

df = pd.DataFrame(data)
df
```
