### NumPy Full Course

NumPy stands for Numerical Python. It is a Python library mainly used for working with numbers, arrays, mathematical calculations, and data efficiently.

The main object in NumPy is called an array. An array is a collection of values stored together.

NumPy arrays are especially useful for:

Mathematical calculations

Large amounts of numerical data

Data analysis

Machine learning

Scientific computing

### Creating a 1D Array

1D means one-dimensional.

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(a)
```

### Creating a 2D Array

2D means rows and columns.

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(a)
```

### Creating a 3D Array

A 3D array contains multiple 2D arrays.

```python
a = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print(a)
```

### Check Dimensions — ndim

ndim tells us how many dimensions an array has.

```python
a = np.array([10, 20, 30])

print(a.ndim)
```

```python
a = np.array([
    [10, 20],
    [30, 40]
])

print(a.ndim)
```

### Shape

shape tells us the number of rows and columns.

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(a.shape)
```

### Size

size tells us the total number of elements.

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(a.size)
```

### Data Type — dtype

dtype tells us the type of values stored in the array.

```python
a = np.array([10, 20, 30])

print(a.dtype)
```

```python
a = np.array([10.5, 20.5, 30.5])

print(a.dtype)
```

### Indexing

NumPy indexing starts from 0.

```python
a = np.array([10, 20, 30, 40])

print(a[0])
print(a[1])
print(a[2])
```

### Negative Indexing

You can access values from the end.

```python
a = np.array([10, 20, 30, 40])

print(a[-1])
```

### Slicing

Slicing selects a range of values.

```python
a = np.array([10, 20, 30, 40, 50])

print(a[1:4])
```

### 2D Array Indexing

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
```

### Creating Arrays with Zeros

np.zeros() creates an array filled with zeros.


```python
a = np.zeros(5)

print(a)
```

```python
a = np.zeros((2, 3))

print(a)
```

### Creating Arrays with Ones

```python
a = np.ones(5)

print(a)
```

```python
a = np.ones((2, 3))

print(a)
```

### arange()

np.arange() creates numbers within a range.

```python
a = np.arange(1, 10)

print(a)
```

```python
a = np.arange(1, 10, 2)

print(a)
```

### linspace()

linspace() creates a specified number of equally spaced values.

```python
a = np.linspace(0, 10, 5)

print(a)
```

### Identity Matrix

np.eye() creates an identity matrix.

```python
a = np.eye(3)

print(a)
```

### Random Numbers

NumPy can generate random numbers.

```python
a = np.random.rand(5)

print(a)
```

```python
a = np.random.randint(1, 10, 5)

print(a)
```

### Arithmetic Operations

One major advantage of NumPy is that we can perform calculations directly on arrays.

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
```

```python
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

### Mathematical Operation

```python
a = np.array([10, 20, 30, 40]) #sum

print(np.sum(a))
```

```python
print(np.mean(a))  #mean
```

```python
print(np.max(a)) # maximum
```

```python
print(np.min(a)) # minimum
```

```python
print(np.std(a))  #standard deviation
```

### Sorting

```python
a = np.array([40, 10, 30, 20])

print(np.sort(a))
```

### Reshape

reshape() changes the shape of an array without changing its values.

```python
a = np.array([1, 2, 3, 4, 5, 6])
```

### Flatten

flatten() converts a multidimensional array into one dimension.

```python
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = a.flatten()

print(b)
```

### Concatenate Arrays

concatenate() joins arrays.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = np.concatenate((a, b))

print(c)
```

### Stack Arrays

vstack() joins arrays vertically.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.vstack((a, b)))
```

### Filtering NumPy Arrays

This is similar to filtering a Pandas DataFrame.

```python
a = np.array([10, 20, 30, 40, 50])

print(a[a > 25])
```
